This is a real decision point, not a minor cleanup, since D15 is one of our two clustering variables, this directly determines how much data you actually get to cluster on.

Let's check if this is a random or a structural pattern (such as survey drop-off, skip logic tied to another question) before deciding how to handle it.

**What's actually going on**

**All 2,268 "missing D15" respondents are missing C1, D11, D12, D13, D14, D15, and D16**, **every single one of them, together, as one block.** And the reverse holds too: every respondent who has D15 also has all the others. This isn't scattered, random missingness, it's a **hard structural break in the survey.**

Combined with the Q07 finding (exactly 2,094 non-null for every Q07 version column, matching precisely the number of respondents *with* D15), the picture is clear:

- **2,094 respondents completed the entire survey**, including the credit score and demographics block at the end.
- **2,268 respondents dropped off partway through** (somewhere after the credit-card behavior questions, before reaching credit score/demographics), and Qualtrics still exported their partial responses as rows, rather than discarding them.

This is a textbook **survey drop-off / incomplete-response pattern**, not a sensitive-question refusal or a data-quality bug. It's also good news methodologically: it means excluding these rows isn't "losing data we could have kept",  it's **correctly excluding incomplete survey responses**, which is standard, well-justified practice, not an arbitrary judgment call.

***Recommended action**

```python
# Keep only respondents who completed the full survey (reached the demographics block)
df_complete = df[df['D15'].notna()].reset_index(drop=True)

print(f"Original: {df.shape[0]} rows")
print(f"Complete responses only: {df_complete.shape[0]} rows")
```

This single filter cleanly resolves missingness in **C1, D11, D12, D13, D14, D15, and D16 all at once**,  not just D15, since they're all missing together. Your cleaning write-up can state this plainly: *"2,268 of 4,362 responses (52%) were partial/incomplete, terminating before the credit score and demographics section; these were excluded as incomplete survey responses, leaving 2,094 complete responses for analysis."* That's a clean, defensible sentence for your report.

**Build the Digital Adoption Score**

```python
digital_cols = ['Q01_8', 'Q01_9', 'Q01_10', 'Q01_11', 'Q01_12', 'Q01_13', 'Q01_14']

# Each column is either "0" (not selected) or the option's text label (selected)
# Binarize: anything other than "0" (and not null) counts as selected
digital_binary = df_complete[digital_cols].apply(lambda col: (col != '0') & col.notna())

# Sum across the 7 columns = digital adoption score, ranging 0-7
df_complete['digital_adoption_score'] = digital_binary.sum(axis=1)

print(df_complete['digital_adoption_score'].value_counts().sort_index())
print(f"\nMean: {df_complete['digital_adoption_score'].mean():.2f}")
```

**Record D15 to an Ordinal Scale**

```python
d15_map = {
    'I do not live paycheck to paycheck': 1,
    'I live paycheck to paycheck but am comfortably paying my bills each month.': 2,
    'I live paycheck to paycheck and have difficulty paying my bills each month.': 3
}

df_complete['D15_ordinal'] = df_complete['D15'].map(d15_map)

# Sanity check — this should show zero unmapped/NaN values, confirming every
# remaining value matched one of our three categories exactly
print(df_complete['D15_ordinal'].value_counts(dropna=False))
print(f"\nUnmapped rows (should be 0): {df_complete['D15_ordinal'].isna().sum()}")
```

Found it, this is a classic type-mismatch bug, not a logic error. The digital adoption score is showing **7 for every single respondent**, which is impossible (that would mean everyone uses all 7 methods), clearly != '0' isn't matching correctly.

**The likely cause:** the "0" values in these columns are probably stored as an actual **numeric 0** (int or float), not the string '0'. When you compare a numeric 0 != '0' (string), Python/pandas treats them as different types and returns `True`, so every "not selected" cell is being incorrectly counted as "selected."

Let's confirm and fix:

```python
# Check the actual dtype and raw values
print(df_complete['Q01_8'].dtype)
print(df_complete['Q01_8'].unique())
```

```python
# Fix: convert to string first, so "0" (string) and 0 (numeric) are handled consistently
digital_cols = ['Q01_8', 'Q01_9', 'Q01_10', 'Q01_11', 'Q01_12', 'Q01_13', 'Q01_14']

digital_binary = df_complete[digital_cols].apply(
    lambda col: (col.astype(str).str.strip() != '0') & col.notna()
)

df_complete['digital_adoption_score'] = digital_binary.sum(axis=1)

print(df_complete['digital_adoption_score'].value_counts().sort_index())
print(f"\nMean: {df_complete['digital_adoption_score'].mean():.2f}")
```

Confirmed, the dtype was `object` but the "not selected" value was stored as a genuine **integer 0**, not the string `'0'` (mixed-type column, common with Excel-sourced data). The `.astype(str)` fix resolved it, and now the distribution looks realistic: mean of 1.37 digital methods used out of 7, with most respondents at 0-2 and a small tail using more, exactly the kind of right-skewed adoption pattern you'd expect for niche digital payment methods.

**Scale Both Variables and Run the Elbow Method**

Now that both variables are clean (`digital_adoption_score:` 0-7, `D15_ordinal`: 1-3), let's standardize them and find the right number of cluster

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

cluster_vars = ['digital_adoption_score', 'D15_ordinal']
X = df_complete[cluster_vars].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Elbow method: fit k-means for k = 1 through 10, track within-cluster sum of squares (WCSS)
wcss = []
k_range = range(1, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    wcss.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.title('Elbow Method for Optimal k')
plt.xticks(k_range)
plt.grid(alpha=0.3)
plt.show()
```

**Silhiuette Score**

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []
k_range_sil = range(2, 11)  # silhouette needs at least 2 clusters

for k in k_range_sil:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"k={k}: silhouette score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(k_range_sil, silhouette_scores, marker='o', color='green')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score by k')
plt.xticks(k_range_sil)
plt.grid(alpha=0.3)
plt.show()
```

**Why silhouette score isn't trustworthy at high k in this specific case**

Your silhouette score **climbs monotonically all the way to k=10** (0.44 then 0.82). Normally a rising silhouette score is a great sign, but here it's actually a red flag, not a recommendation to use k=10. Here's why: your two clustering variables have very limited possible values, `digital_adoption_score` (0-7) and `D15_ordinal` (1-3), meaning there are at most **24 unique (x, y)** combinations in the entire dataset. As k approaches that number, k-means starts carving out clusters around individual data points/tiny combinations rather than finding genuine behavioral groups. Silhouette score trivially improves as clusters shrink toward single points, but that's overfitting to the grid of discrete values, not meaningful segmentation. **A 10-segment marketing strategy built on 2 low-cardinality variables would not be defensible or actionable.**

**What the elbow plot actually shows**

The elbow plot is much more trustworthy here, and it shows a clear bend: **WCSS drops sharply from k=1 to k=3, then flattens out, the marginal improvement from k=4 onward is small.** This matches the silhouette pattern too, if you look at just the low end: scores rise meaningfully from k=2 (0.44) then k=3 (0.51) then k=4 (0.57), but then barely move from k=4 to k=5 (0.5678 then 0.5700, essentially flat) before climbing again, and that later climb is the overfitting effect described above, not real signal.

**Recommendation: k = 4**

This is the point where the elbow bends and where silhouette score plateaus before the artificial climb, a defensible, "real" choice rather than a runaway high-k result. k = 3 is also reasonable if you want a simpler story (and notably lines up with D15's 3 natural categories).

```python
# Fit final model with k=4
kmeans_final = KMeans(n_clusters=4, random_state=42, n_init=10)
df_complete['cluster'] = kmeans_final.fit_predict(X_scaled)

print(df_complete['cluster'].value_counts().sort_index())
```

The elbow method showed a clear bend at k=3–4, with diminishing reductions in within-cluster sum of squares beyond that point. While silhouette scores continued to rise at higher values of k, this reflects the limited number of unique value combinations in the two clustering variables (digital adoption score: 0–7; financial lifestyle: 1–3 ordinal levels) rather than genuinely improved segmentation,  at high k, clusters begin to form around individual value combinations rather than meaningful behavioral groups. Silhouette scores plateaued between k=4 and k=5 (0.568 vs. 0.570), reinforcing k=4 as a well-supported, interpretable choice.
