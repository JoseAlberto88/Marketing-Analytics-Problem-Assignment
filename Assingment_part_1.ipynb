```python
import pandas as pd

file_path = '/content/drive/MyDrive/Marketing analytics/Purchasing_data.xlsx'

# Read with no header assumed, so we can control exactly which rows are what
df_raw = pd.read_excel(file_path, header=None)

# Row 0 = short codes -> use these as the actual column names
short_codes = df_raw.iloc[0]

# Row 1 = long question text -> save as a lookup dictionary (short code -> full question)
question_labels = dict(zip(short_codes, df_raw.iloc[1]))

# Row 2 onward = actual respondent data
df = df_raw.iloc[2:].reset_index(drop=True)
df.columns = short_codes

print(f"Fixed shape: {df.shape}")
print("\nExample lookup:")
print("D15 =", question_labels["D15"])

df.head()
```

```python
# ============================================================
# Full column map: short code -> full question text
# ============================================================
import pandas as pd

column_overview = pd.DataFrame({
    "Column Code": df.columns,
    "Question Text": [question_labels.get(col, "N/A") for col in df.columns]
})

pd.set_option('display.max_colwidth', None)
display(column_overview)
```

```python
for _, row in column_overview.iterrows():
    print(f"{row['Column Code']}: {row['Question Text']}")
```

Let's now inspect teh actual values in the columns we need, using the `df` dataset

```python
# Listing the columns that we are interesting in

digital_cols = ['Q01_8', 'Q01_9', 'Q01_10', 'Q01_11', 'Q01_12', 'Q01_13', 'Q01_14']

for col in digital_cols:
    print(f"--- {col}: {question_labels[col]} ---")
    print(df[col].value_counts(dropna=False))
    print()

print("--- D15:", question_labels['D15'], "---")
print(df['D15'].value_counts(dropna=False))
```

**What the data actually shows**

**Digital payment columns (Q01_8–Q01_14):** Clean pattern,  each is either the literal string `"0"` (not selected) or the option's text label (selected). Easy to binarize. **One data-quality wrinkle:** Q01_8 has inconsistent capitalization , "PayPal" (2,119) vs. "Paypal" (46) are the same answer, just typed differently by Qualtrics/respondents. Doesn't affect our binary flag (both just mean "selected"), but worth noting in your cleaning write-up as an inconsistent-value issue you caught and handled.

**D15 (financial lifestyle):** Three genuine ordinal categories, but a big problem: **2,268 of 4,362 respondents (52%) have no answer at all.** That's more than half your sample. This is a real decision point, not a minor cleanup, since D15 is one of our two clustering variables, this directly determines how much data you actually get to cluster on.

Let's check if this is a random or a structural pattern (such as survey drop-off, skip logic tied to another question) before deciding how to handle it.

```python
# 1. Check if there are extra Qualtrics metadata columns not in the codebook
# (e.g., Progress, Finished, Duration) that might explain drop-off
print(df.columns.tolist())
```

```python
# 2. Does D15 missingness line up with missingness in nearby demographic columns?
# If respondents who skipped D15 also skipped D16, D14, D13, etc., it suggests
# survey abandonment (they stopped before reaching the end), not a D15-specific issue.
nearby_cols = ['C1', 'D11', 'D12', 'D13', 'D14', 'D15', 'D16']

missing_pattern = df[nearby_cols].isna()
print(missing_pattern.sum())
print()

# Cross-tab: of those missing D15, how many are ALSO missing D16 (the very last question)?
d15_missing = df['D15'].isna()
print("Among respondents missing D15:")
print(df.loc[d15_missing, 'D16'].isna().value_counts())
print()
print("Among respondents NOT missing D15:")
print(df.loc[~d15_missing, 'D16'].isna().value_counts())
```

```python
# 3. Check if there's a Progress/Finished/Duration column (common in raw Qualtrics exports)
# that directly tells us who completed the survey vs. dropped off partway
possible_meta_cols = [c for c in df.columns if any(k in str(c).lower() for k in ['progress', 'finish', 'duration', 'status'])]
print("Possible metadata columns found:", possible_meta_cols)

if possible_meta_cols:
    for col in possible_meta_cols:
        print(f"\n--- {col} ---")
        print(df[col].value_counts(dropna=False).head(10))
```

```python
# 4. Alternative check: does D15 missingness correlate with which Q07/Q12 "version"
# a respondent got? (Since this survey clearly has multiple randomized question-wording
# versions, e.g. Q07_1 through Q07_6, Q12_1 through Q12_9 — a respondent likely only
# answers ONE version, leaving the rest blank by design, not by error.)
q07_cols = ['Q07_1', 'Q07_2', 'Q07_3', 'Q07_4', 'Q07_5', 'Q07_6']
existing_q07 = [c for c in q07_cols if c in df.columns]
print(df[existing_q07].notna().sum())
```

**What's actually going on**

**All 2,268 "missing D15" respondents are missing C1, D11, D12, D13, D14, D15, and D16**, **every single one of them, together, as one block.** And the reverse holds too: every respondent who has D15 also has all the others. This isn't scattered, random missingness, it's a **hard structural break in the survey.**

Combined with the Q07 finding (exactly 2,094 non-null for every Q07 version column, matching precisely the number of respondents *with* D15), the picture is clear:

- **2,094 respondents completed the entire survey**, including the credit score and demographics block at the end.
- **2,268 respondents dropped off partway through** (somewhere after the credit-card behavior questions, before reaching credit score/demographics), and Qualtrics still exported their partial responses as rows, rather than discarding them.

This is a textbook **survey drop-off / incomplete-response pattern**, not a sensitive-question refusal or a data-quality bug. It's also good news methodologically: it means excluding these rows isn't "losing data we could have kept",  it's **correctly excluding incomplete survey responses**, which is standard, well-justified practice, not an arbitrary judgment call.

***Recommended action**

```python
# Keep only respondents who completed the full survey (reached the demographics block)
df_complete = df[df['D15'].notna()].reset_index(drop=True)

print(f"Original: {df.shape[0]} rows")
print(f"Complete responses only: {df_complete.shape[0]} rows")
```

This single filter cleanly resolves missingness in **C1, D11, D12, D13, D14, D15, and D16 all at once**,  not just D15, since they're all missing together. Your cleaning write-up can state this plainly: *"2,268 of 4,362 responses (52%) were partial/incomplete, terminating before the credit score and demographics section; these were excluded as incomplete survey responses, leaving 2,094 complete responses for analysis."* That's a clean, defensible sentence for your report.
